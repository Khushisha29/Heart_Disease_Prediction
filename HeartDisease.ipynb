# Train Test Split

```python
X = df.drop(['target'], axis = 1)
y = df['target']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 0)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
model_lg = LogisticRegression(random_state=0)
model_lg.fit(X_train, y_train)
```

```python
pred_lg = model_lg.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
lg = accuracy_score(y_test, pred_lg)
print(lg)
```

```python
print(classification_report(y_test,pred_lg))
```

```python
cm_1 = confusion_matrix(y_test, pred_lg)
ax = plt.subplot()
sns.heatmap(cm_1, annot=True, ax=ax, cmap='Blues', fmt='g')
ax.set_xlabel('Predicted Labels')
ax.set_ylabel('True Labels')
ax.set_title('Logistic Regression: Confusion Matrix')
```

```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

fpr_lr, tpr_lr, _ = roc_curve(y_test, pred_lg)
plt.plot(fpr_lr, tpr_lr, label='Logistic Regression')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend()
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

clf = RandomForestClassifier(n_estimators=100, random_state=42)

# Fit the model on the data
clf.fit(X, y)

# Get feature importances from the trained model
feature_importances = clf.feature_importances_

# Create a DataFrame to store feature importances
importances_df = pd.DataFrame({'Feature': X.columns, 'Importance': feature_importances})

# Sort the DataFrame by importance in descending order
importances_df = importances_df.sort_values(by='Importance', ascending=False)

# Plot feature importances
plt.figure(figsize=(10, 6))
plt.bar(importances_df['Feature'], importances_df['Importance'])
plt.xticks(rotation=90)
plt.title('Feature Importance from Random Forest')
plt.xlabel('Feature')
plt.ylabel('Importance')
plt.tight_layout()
plt.show()
```
